```python
# ============================================================
# STEP 1: CSV UPLOAD, DATA LOADING & INITIAL INSPECTION
# Research-Oriented Machine Learning Workflow
# ============================================================

from google.colab import files
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. Upload CSV file
# ------------------------------------------------------------
uploaded = files.upload()

# ------------------------------------------------------------
# 2. Get uploaded file name
# ------------------------------------------------------------
filename = next(iter(uploaded))

print("=" * 70)
print("CSV FILE UPLOADED SUCCESSFULLY")
print("=" * 70)

print("File name:", filename)

# ------------------------------------------------------------
# 3. Load dataset
# ------------------------------------------------------------
df = pd.read_csv(filename)

# ------------------------------------------------------------
# 4. Dataset dimensions
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("DATASET DIMENSION")
print("=" * 70)

print("Number of rows    :", df.shape[0])
print("Number of columns :", df.shape[1])

# ------------------------------------------------------------
# 5. Column names
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("COLUMN NAMES")
print("=" * 70)

for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")

# ------------------------------------------------------------
# 6. First 5 observations
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("FIRST 5 ROWS")
print("=" * 70)

display(df.head())

# ------------------------------------------------------------
# 7. Last 5 observations
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("LAST 5 ROWS")
print("=" * 70)

display(df.tail())

# ------------------------------------------------------------
# 8. Data types
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("DATA TYPES")
print("=" * 70)

display(df.dtypes.to_frame(name="Data Type"))

# ------------------------------------------------------------
# 9. Dataset information
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

df.info()

# ------------------------------------------------------------
# 10. Statistical summary
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("STATISTICAL SUMMARY")
print("=" * 70)

display(df.describe(include="all").T)

# ------------------------------------------------------------
# 11. Target column verification
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("TARGET COLUMN CHECK")
print("=" * 70)

if "disease_label" in df.columns:
    print("✓ 'disease_label' column found.")
    print("Target data type:", df["disease_label"].dtype)
else:
    print("⚠ WARNING: 'disease_label' column was NOT found.")
    print("Please check the target column name.")
```

```python
# ============================================================
# STEP 2: MISSING VALUES AND DUPLICATE CHECK
# Research-Oriented ML Workflow
# ============================================================

# ------------------------------------------------------------
# 1. Check missing values
# ------------------------------------------------------------
print("=" * 70)
print("MISSING VALUE ANALYSIS")
print("=" * 70)

missing_count = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_table = pd.DataFrame({
    "Missing Values": missing_count,
    "Missing Percentage (%)": missing_percentage
})

display(missing_table)

# ------------------------------------------------------------
# 2. Total missing values
# ------------------------------------------------------------
total_missing = df.isnull().sum().sum()

print("\nTotal missing values in dataset:", total_missing)

if total_missing == 0:
    print("✓ No missing values detected.")
else:
    print("⚠ Missing values detected. Cleaning will be required.")

# ------------------------------------------------------------
# 3. Check duplicate rows
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("DUPLICATE ROW ANALYSIS")
print("=" * 70)

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

if duplicate_count == 0:
    print("✓ No duplicate rows detected.")
else:
    print("⚠ Duplicate rows detected.")

# ------------------------------------------------------------
# 4. Duplicate percentage
# ------------------------------------------------------------
duplicate_percentage = (duplicate_count / len(df)) * 100

print("Duplicate percentage: {:.2f}%".format(duplicate_percentage))

# ------------------------------------------------------------
# 5. Display duplicate rows if present
# ------------------------------------------------------------
if duplicate_count > 0:
    print("\nDuplicate rows:")
    display(df[df.duplicated(keep=False)].sort_values(by=df.columns.tolist()))
else:
    print("\nNo duplicate rows to display.")

# ------------------------------------------------------------
# 6. Check unique values of categorical columns
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("CATEGORICAL FEATURE VALUE CHECK")
print("=" * 70)

categorical_columns = df.select_dtypes(include=["object", "category"]).columns

for column in categorical_columns:
    print(f"\n{column}:")
    print("Number of unique values:", df[column].nunique())
    print("Unique values:", df[column].unique())

# ------------------------------------------------------------
# 7. Target variable distribution
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("TARGET VARIABLE: disease_label")
print("=" * 70)

print("Number of unique disease classes:",
      df["disease_label"].nunique())

print("\nDisease classes:")
print(df["disease_label"].unique())

print("\nDisease label frequency:")
display(df["disease_label"].value_counts())

print("\nDisease label percentage:")
display(
    df["disease_label"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
    .rename("Percentage (%)")
)
```

```python
# ============================================================
# STEP 3: DATA CLEANING & PREPROCESSING
# Research-Oriented ML Workflow
# ============================================================

print("=" * 70)
print("STEP 3: DATA CLEANING AND PREPROCESSING")
print("=" * 70)


# ------------------------------------------------------------
# 1. Create a copy of the original dataset
# ------------------------------------------------------------
df_clean = df.copy()

print("\nOriginal dataset shape:", df_clean.shape)


# ------------------------------------------------------------
# 2. Check column names for leading/trailing spaces
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("COLUMN NAME CLEANING")
print("=" * 70)

df_clean.columns = df_clean.columns.str.strip()

print("Cleaned column names:")
print(df_clean.columns.tolist())


# ------------------------------------------------------------
# 3. Check categorical columns for leading/trailing spaces
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("CATEGORICAL VALUE CLEANING")
print("=" * 70)

categorical_columns = ["crop_type", "season", "disease_label"]

for column in categorical_columns:

    # Check values before stripping
    original_values = df_clean[column].unique()

    # Remove leading/trailing whitespace
    df_clean[column] = df_clean[column].astype(str).str.strip()

    cleaned_values = df_clean[column].unique()

    print(f"\n{column}")
    print("Unique values after cleaning:")
    print(cleaned_values)


# ------------------------------------------------------------
# 4. Check empty strings
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("EMPTY STRING CHECK")
print("=" * 70)

empty_string_counts = {}

for column in df_clean.columns:
    empty_string_counts[column] = (
        df_clean[column]
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

empty_string_table = pd.DataFrame.from_dict(
    empty_string_counts,
    orient="index",
    columns=["Empty String Count"]
)

display(empty_string_table)


# ------------------------------------------------------------
# 5. Numerical feature identification
# ------------------------------------------------------------
numeric_columns = df_clean.select_dtypes(
    include=["int64", "float64", "int32", "float32"]
).columns.tolist()

print("\n" + "=" * 70)
print("NUMERICAL FEATURES")
print("=" * 70)

print(numeric_columns)


# ------------------------------------------------------------
# 6. Check for infinite values
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("INFINITE VALUE CHECK")
print("=" * 70)

infinite_counts = np.isinf(
    df_clean[numeric_columns]
).sum()

display(infinite_counts.to_frame(name="Infinite Values"))

if infinite_counts.sum() == 0:
    print("✓ No infinite values detected.")
else:
    print("⚠ Infinite values detected.")


# ------------------------------------------------------------
# 7. Check numerical ranges
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("NUMERICAL FEATURE RANGE CHECK")
print("=" * 70)

range_table = pd.DataFrame({
    "Minimum": df_clean[numeric_columns].min(),
    "Maximum": df_clean[numeric_columns].max(),
    "Mean": df_clean[numeric_columns].mean(),
    "Std": df_clean[numeric_columns].std()
})

display(range_table)


# ------------------------------------------------------------
# 8. Check for constant features
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("CONSTANT FEATURE CHECK")
print("=" * 70)

unique_counts = df_clean.nunique()

constant_columns = unique_counts[
    unique_counts <= 1
].index.tolist()

if len(constant_columns) == 0:
    print("✓ No constant features detected.")
else:
    print("⚠ Constant columns detected:")
    print(constant_columns)


# ------------------------------------------------------------
# 9. Statistical outlier detection using IQR
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("OUTLIER ANALYSIS USING IQR")
print("=" * 70)

outlier_summary = []

for column in numeric_columns:

    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (df_clean[column] < lower_bound) |
        (df_clean[column] > upper_bound)
    ).sum()

    outlier_percentage = (
        outlier_count / len(df_clean)
    ) * 100

    outlier_summary.append({
        "Feature": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": outlier_count,
        "Outlier Percentage (%)": outlier_percentage
    })

outlier_table = pd.DataFrame(outlier_summary)

display(outlier_table)


# ------------------------------------------------------------
# 10. Final duplicate check after cleaning
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("POST-CLEANING DUPLICATE CHECK")
print("=" * 70)

print(
    "Duplicate rows after cleaning:",
    df_clean.duplicated().sum()
)


# ------------------------------------------------------------
# 11. Final missing-value check
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("POST-CLEANING MISSING VALUE CHECK")
print("=" * 70)

print(
    "Total missing values after cleaning:",
    df_clean.isnull().sum().sum()
)


# ------------------------------------------------------------
# 12. Final dataset structure
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("FINAL CLEANED DATASET STRUCTURE")
print("=" * 70)

print("Rows:", df_clean.shape[0])
print("Columns:", df_clean.shape[1])

display(df_clean.head())
```

```python
# ============================================================
# STEP 4: ENCODE CATEGORICAL FEATURES
# Research-Oriented ML Workflow
# ============================================================

from sklearn.preprocessing import OneHotEncoder

print("=" * 70)
print("STEP 4: CATEGORICAL FEATURE ENCODING")
print("=" * 70)


# ------------------------------------------------------------
# 1. Identify categorical feature columns
# ------------------------------------------------------------

categorical_features = ["crop_type", "season"]

print("\nCategorical features:")
print(categorical_features)


# ------------------------------------------------------------
# 2. Check their unique categories
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CATEGORICAL FEATURE CATEGORIES")
print("=" * 70)

for column in categorical_features:
    print(f"\n{column}:")
    print(df_clean[column].unique())


# ------------------------------------------------------------
# 3. One-Hot Encode categorical features
# ------------------------------------------------------------

encoder = OneHotEncoder(
    sparse_output=False,
    handle_unknown="ignore"
)

encoded_array = encoder.fit_transform(
    df_clean[categorical_features]
)


# ------------------------------------------------------------
# 4. Get encoded feature names
# ------------------------------------------------------------

encoded_feature_names = encoder.get_feature_names_out(
    categorical_features
)

print("\n" + "=" * 70)
print("ENCODED FEATURE NAMES")
print("=" * 70)

print(encoded_feature_names.tolist())


# ------------------------------------------------------------
# 5. Create encoded DataFrame
# ------------------------------------------------------------

encoded_df = pd.DataFrame(
    encoded_array,
    columns=encoded_feature_names,
    index=df_clean.index
)


# ------------------------------------------------------------
# 6. Select numerical features
# ------------------------------------------------------------

numerical_features = [
    "temperature",
    "humidity",
    "rainfall",
    "soil_moisture",
    "soil_ph",
    "light_intensity"
]


# ------------------------------------------------------------
# 7. Combine numerical + encoded categorical features
# ------------------------------------------------------------

X_encoded = pd.concat(
    [
        df_clean[numerical_features],
        encoded_df
    ],
    axis=1
)


# ------------------------------------------------------------
# 8. Display final feature matrix
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("ENCODED FEATURE MATRIX")
print("=" * 70)

print("Number of samples:", X_encoded.shape[0])
print("Number of features:", X_encoded.shape[1])

display(X_encoded.head())


# ------------------------------------------------------------
# 9. Check data types after encoding
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FEATURE DATA TYPES AFTER ENCODING")
print("=" * 70)

display(X_encoded.dtypes.to_frame(name="Data Type"))


# ------------------------------------------------------------
# 10. Check missing values after encoding
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("MISSING VALUE CHECK AFTER ENCODING")
print("=" * 70)

print(
    "Total missing values:",
    X_encoded.isnull().sum().sum()
)
```

```python
# ============================================================
# STEP 5: SEPARATE FEATURES (X) AND TARGET (y)
# Research-Oriented ML Workflow
# ============================================================

print("=" * 70)
print("STEP 5: FEATURE-TARGET SEPARATION")
print("=" * 70)


# ------------------------------------------------------------
# 1. Define feature matrix X
# ------------------------------------------------------------

X = X_encoded.copy()


# ------------------------------------------------------------
# 2. Define target variable y
# ------------------------------------------------------------

y = df_clean["disease_label"].copy()


# ------------------------------------------------------------
# 3. Display feature information
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FEATURE MATRIX (X)")
print("=" * 70)

print("X shape:", X.shape)

print("\nFeature names:")
for i, feature in enumerate(X.columns, start=1):
    print(f"{i}. {feature}")


# ------------------------------------------------------------
# 4. Display target information
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TARGET VARIABLE (y)")
print("=" * 70)

print("y shape:", y.shape)

print("\nTarget variable name:")
print(y.name)

print("\nTarget data type:")
print(y.dtype)


# ------------------------------------------------------------
# 5. Display target classes
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TARGET CLASSES")
print("=" * 70)

target_classes = y.unique()

print("Number of classes:", len(target_classes))

print("\nClasses:")
for i, label in enumerate(target_classes, start=1):
    print(f"{i}. {label}")


# ------------------------------------------------------------
# 6. Verify X and y have the same number of observations
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("X-y DIMENSION CONSISTENCY CHECK")
print("=" * 70)

print("Number of X observations:", len(X))
print("Number of y observations:", len(y))

if len(X) == len(y):
    print("✓ X and y have the same number of observations.")
else:
    print("⚠ ERROR: X and y have different numbers of observations.")


# ------------------------------------------------------------
# 7. Check missing values in X and y
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("X-y MISSING VALUE CHECK")
print("=" * 70)

print("Missing values in X:", X.isnull().sum().sum())
print("Missing values in y:", y.isnull().sum())


# ------------------------------------------------------------
# 8. Display first 5 rows of X
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FIRST 5 ROWS OF X")
print("=" * 70)

display(X.head())


# ------------------------------------------------------------
# 9. Display first 5 values of y
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FIRST 5 VALUES OF y")
print("=" * 70)

display(y.head())
```

```python
# ============================================================
# STEP 6: CLASS IMBALANCE ANALYSIS
# Research-Oriented ML Workflow
# ============================================================

import matplotlib.pyplot as plt

print("=" * 70)
print("STEP 6: CLASS IMBALANCE ANALYSIS")
print("=" * 70)


# ------------------------------------------------------------
# 1. Calculate class frequency
# ------------------------------------------------------------

class_counts = y.value_counts()

print("\n" + "=" * 70)
print("DISEASE CLASS FREQUENCY")
print("=" * 70)

print(class_counts)


# ------------------------------------------------------------
# 2. Calculate class percentage
# ------------------------------------------------------------

class_percentages = (
    y.value_counts(normalize=True) * 100
).round(2)

print("\n" + "=" * 70)
print("DISEASE CLASS PERCENTAGE")
print("=" * 70)

print(class_percentages)


# ------------------------------------------------------------
# 3. Create research-friendly class distribution table
# ------------------------------------------------------------

class_distribution = pd.DataFrame({
    "Disease Class": class_counts.index,
    "Frequency": class_counts.values,
    "Percentage (%)": class_percentages.values
})

print("\n" + "=" * 70)
print("CLASS DISTRIBUTION TABLE")
print("=" * 70)

display(class_distribution)


# ------------------------------------------------------------
# 4. Calculate imbalance ratio
# ------------------------------------------------------------

largest_class = class_counts.max()
smallest_class = class_counts.min()

imbalance_ratio = largest_class / smallest_class

print("\n" + "=" * 70)
print("IMBALANCE RATIO")
print("=" * 70)

print("Largest class size :", largest_class)
print("Smallest class size:", smallest_class)
print("Imbalance ratio    : {:.2f}".format(imbalance_ratio))


# ------------------------------------------------------------
# 5. Identify majority and minority classes
# ------------------------------------------------------------

majority_class = class_counts.idxmax()
minority_class = class_counts.idxmin()

print("\nMajority class:", majority_class)
print("Minority class:", minority_class)


# ------------------------------------------------------------
# 6. Basic imbalance interpretation
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("IMBALANCE INTERPRETATION")
print("=" * 70)

if imbalance_ratio < 1.5:
    print("The dataset shows relatively low class imbalance.")
elif imbalance_ratio < 3:
    print("The dataset shows moderate class imbalance.")
else:
    print("The dataset shows substantial class imbalance.")


# ------------------------------------------------------------
# 7. Plot class distribution
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.bar(
    class_distribution["Disease Class"],
    class_distribution["Frequency"]
)

plt.xlabel("Disease Class")
plt.ylabel("Number of Samples")
plt.title("Distribution of Disease Classes")

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 8. Verify total observations
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TOTAL SAMPLE VERIFICATION")
print("=" * 70)

print("Total samples from X:", len(X))
print("Total samples from y:", len(y))
print("Sum of class frequencies:", class_counts.sum())

if class_counts.sum() == len(y):
    print("✓ Class frequencies correctly account for all observations.")
else:
    print("⚠ ERROR: Class frequencies do not match total observations.")
```

```python
# ============================================================
# STEP 7: STRATIFIED TRAIN-TEST SPLIT
# Research-Oriented ML Workflow
# ============================================================

from sklearn.model_selection import train_test_split

print("=" * 70)
print("STEP 7: STRATIFIED TRAIN-TEST SPLIT")
print("=" * 70)


# ------------------------------------------------------------
# 1. Perform stratified train-test split
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


# ------------------------------------------------------------
# 2. Display dataset sizes
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("DATASET SPLIT SUMMARY")
print("=" * 70)

print("Original dataset size :", len(X))
print("Training set size     :", len(X_train))
print("Testing set size      :", len(X_test))


# ------------------------------------------------------------
# 3. Display feature dimensions
# ------------------------------------------------------------

print("\nTraining feature shape:", X_train.shape)
print("Testing feature shape :", X_test.shape)

print("\nTraining target shape:", y_train.shape)
print("Testing target shape :", y_test.shape)


# ------------------------------------------------------------
# 4. Check class distribution in training set
# ------------------------------------------------------------

train_distribution = pd.DataFrame({
    "Training Count": y_train.value_counts(),
    "Training Percentage (%)":
        y_train.value_counts(normalize=True).mul(100).round(2)
})

print("\n" + "=" * 70)
print("TRAINING SET CLASS DISTRIBUTION")
print("=" * 70)

display(train_distribution)


# ------------------------------------------------------------
# 5. Check class distribution in testing set
# ------------------------------------------------------------

test_distribution = pd.DataFrame({
    "Testing Count": y_test.value_counts(),
    "Testing Percentage (%)":
        y_test.value_counts(normalize=True).mul(100).round(2)
})

print("\n" + "=" * 70)
print("TESTING SET CLASS DISTRIBUTION")
print("=" * 70)

display(test_distribution)


# ------------------------------------------------------------
# 6. Compare original, training and testing proportions
# ------------------------------------------------------------

original_percentage = (
    y.value_counts(normalize=True)
    .mul(100)
    .round(2)
)

train_percentage = (
    y_train.value_counts(normalize=True)
    .mul(100)
    .round(2)
)

test_percentage = (
    y_test.value_counts(normalize=True)
    .mul(100)
    .round(2)
)

comparison_table = pd.DataFrame({
    "Original (%)": original_percentage,
    "Training (%)": train_percentage,
    "Testing (%)": test_percentage
})

print("\n" + "=" * 70)
print("CLASS PROPORTION COMPARISON")
print("=" * 70)

display(comparison_table)


# ------------------------------------------------------------
# 7. Verify no index overlap between train and test
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TRAIN-TEST OVERLAP CHECK")
print("=" * 70)

overlap = set(X_train.index).intersection(set(X_test.index))

print("Number of overlapping observations:", len(overlap))

if len(overlap) == 0:
    print("✓ No observation overlap between training and testing sets.")
else:
    print("⚠ WARNING: Overlap detected.")


# ------------------------------------------------------------
# 8. Verify stratification
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("STRATIFICATION CHECK")
print("=" * 70)

max_difference = (
    comparison_table["Training (%)"] -
    comparison_table["Testing (%)"]
).abs().max()

print(
    "Maximum percentage-point difference "
    "between training and testing:",
    round(max_difference, 2)
)

print("✓ Stratified split completed successfully.")
```

```python
# ============================================================
# STEP 8: RANDOM FOREST MODEL TRAINING
# Research-Oriented ML Workflow
# ============================================================

from sklearn.ensemble import RandomForestClassifier

print("=" * 70)
print("STEP 8: RANDOM FOREST MODEL TRAINING")
print("=" * 70)


# ------------------------------------------------------------
# 1. Initialize Random Forest classifier
# ------------------------------------------------------------

rf_model = RandomForestClassifier(
    n_estimators=500,
    criterion="gini",
    max_features="sqrt",
    class_weight="balanced",
    random_state=42,
    n_jobs=-1,
    oob_score=True
)


# ------------------------------------------------------------
# 2. Train Random Forest
# ------------------------------------------------------------

print("\nTraining Random Forest...")

rf_model.fit(X_train, y_train)

print("✓ Random Forest training completed successfully.")


# ------------------------------------------------------------
# 3. Display model configuration
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("RANDOM FOREST CONFIGURATION")
print("=" * 70)

print("Number of trees       :", rf_model.n_estimators)
print("Criterion             :", rf_model.criterion)
print("Max features          :", rf_model.max_features)
print("Class weighting       :", rf_model.class_weight)
print("Random state          :", rf_model.random_state)
print("Number of features    :", X_train.shape[1])
print("Number of training samples:", X_train.shape[0])


# ------------------------------------------------------------
# 4. Out-of-Bag (OOB) score
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("OUT-OF-BAG (OOB) SCORE")
print("=" * 70)

print(
    "Random Forest OOB Score: {:.4f}".format(
        rf_model.oob_score_
    )
)


# ------------------------------------------------------------
# 5. Verify trained model
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("MODEL VERIFICATION")
print("=" * 70)

print("✓ Model object created :", type(rf_model).__name__)
print("✓ Number of estimators :", len(rf_model.estimators_))
print("✓ Model is ready for evaluation.")
```

```python
# ============================================================
# STEP 14: MODEL VALIDATION / RELIABILITY
# Research-Oriented ML Workflow
#
# Covers:
#   - Stratified K-Fold Cross-Validation (RF + XGBoost)
#   - Macro F1 mean +/- standard deviation
#   - Overfitting check (Train-Validation gap)
#   - Held-out test vs CV consistency check
#   - Statistical significance test (Wilcoxon signed-rank)
# ============================================================

from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import f1_score, accuracy_score
from sklearn.base import clone
from sklearn.utils.class_weight import compute_sample_weight
from scipy import stats
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("=" * 70)
print("STEP 14: MODEL VALIDATION / RELIABILITY")
print("=" * 70)


# ------------------------------------------------------------
# 1. Define Stratified K-Fold cross-validation strategy
# ------------------------------------------------------------

N_SPLITS = 5

cv = StratifiedKFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=42
)

print("\nCross-validation strategy : Stratified K-Fold")
print("Number of folds           :", N_SPLITS)
print("Performed on              : Training set only (X_train, y_train)")
print("Test set                  : Untouched (held out)")


# ------------------------------------------------------------
# 2. Random Forest -- manual Stratified K-Fold cross-validation
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("RANDOM FOREST -- CROSS-VALIDATION")
print("=" * 70)

rf_cv_train_f1 = []
rf_cv_val_f1 = []
rf_cv_val_acc = []

for fold, (train_idx, val_idx) in enumerate(cv.split(X_train, y_train), start=1):

    X_tr_fold = X_train.iloc[train_idx]
    y_tr_fold = y_train.iloc[train_idx]

    X_val_fold = X_train.iloc[val_idx]
    y_val_fold = y_train.iloc[val_idx]

    rf_fold_model = clone(rf_model)
    rf_fold_model.fit(X_tr_fold, y_tr_fold)

    # Performance on the fold's own training data
    y_tr_pred = rf_fold_model.predict(X_tr_fold)
    train_f1 = f1_score(y_tr_fold, y_tr_pred, average="macro", zero_division=0)

    # Performance on the held-out validation fold
    y_val_pred = rf_fold_model.predict(X_val_fold)
    val_f1 = f1_score(y_val_fold, y_val_pred, average="macro", zero_division=0)
    val_acc = accuracy_score(y_val_fold, y_val_pred)

    rf_cv_train_f1.append(train_f1)
    rf_cv_val_f1.append(val_f1)
    rf_cv_val_acc.append(val_acc)

    print(
        f"Fold {fold}: "
        f"Train Macro F1 = {train_f1:.4f} | "
        f"Val Macro F1 = {val_f1:.4f} | "
        f"Val Accuracy = {val_acc:.4f}"
    )

rf_cv_train_f1 = np.array(rf_cv_train_f1)
rf_cv_val_f1 = np.array(rf_cv_val_f1)
rf_cv_val_acc = np.array(rf_cv_val_acc)


# ------------------------------------------------------------
# 3. Random Forest -- cross-validation summary
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("RANDOM FOREST -- CROSS-VALIDATION SUMMARY")
print("=" * 70)

print("Macro F1 (Train folds)     : {:.4f} +/- {:.4f}".format(
    rf_cv_train_f1.mean(), rf_cv_train_f1.std()
))
print("Macro F1 (Validation folds): {:.4f} +/- {:.4f}".format(
    rf_cv_val_f1.mean(), rf_cv_val_f1.std()
))
print("Accuracy (Validation folds): {:.4f} +/- {:.4f}".format(
    rf_cv_val_acc.mean(), rf_cv_val_acc.std()
))

rf_overfit_gap = rf_cv_train_f1.mean() - rf_cv_val_f1.mean()
print("\nTrain-Validation Macro F1 Gap (Overfitting indicator): {:.4f}".format(
    rf_overfit_gap
))


# ------------------------------------------------------------
# 4. XGBoost -- manual Stratified K-Fold cross-validation
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("XGBOOST -- CROSS-VALIDATION")
print("=" * 70)

xgb_cv_train_f1 = []
xgb_cv_val_f1 = []
xgb_cv_val_acc = []

# Encode the full training target once (consistent with label_encoder from Step 10)
y_train_encoded_full = label_encoder.transform(y_train)

for fold, (train_idx, val_idx) in enumerate(cv.split(X_train, y_train), start=1):

    X_tr_fold = X_train.iloc[train_idx]
    y_tr_fold_enc = y_train_encoded_full[train_idx]

    X_val_fold = X_train.iloc[val_idx]
    y_val_fold_enc = y_train_encoded_full[val_idx]

    # Balanced sample weights computed ONLY from this fold's training data
    fold_sample_weights = compute_sample_weight(
        class_weight="balanced",
        y=y_tr_fold_enc
    )

    xgb_fold_model = clone(xgb_model)
    xgb_fold_model.fit(
        X_tr_fold,
        y_tr_fold_enc,
        sample_weight=fold_sample_weights
    )

    # Performance on the fold's own training data
    y_tr_pred_enc = xgb_fold_model.predict(X_tr_fold)
    train_f1 = f1_score(y_tr_fold_enc, y_tr_pred_enc, average="macro", zero_division=0)

    # Performance on the held-out validation fold
    y_val_pred_enc = xgb_fold_model.predict(X_val_fold)
    val_f1 = f1_score(y_val_fold_enc, y_val_pred_enc, average="macro", zero_division=0)
    val_acc = accuracy_score(y_val_fold_enc, y_val_pred_enc)

    xgb_cv_train_f1.append(train_f1)
    xgb_cv_val_f1.append(val_f1)
    xgb_cv_val_acc.append(val_acc)

    print(
        f"Fold {fold}: "
        f"Train Macro F1 = {train_f1:.4f} | "
        f"Val Macro F1 = {val_f1:.4f} | "
        f"Val Accuracy = {val_acc:.4f}"
    )

xgb_cv_train_f1 = np.array(xgb_cv_train_f1)
xgb_cv_val_f1 = np.array(xgb_cv_val_f1)
xgb_cv_val_acc = np.array(xgb_cv_val_acc)


# ------------------------------------------------------------
# 5. XGBoost -- cross-validation summary
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("XGBOOST -- CROSS-VALIDATION SUMMARY")
print("=" * 70)

print("Macro F1 (Train folds)     : {:.4f} +/- {:.4f}".format(
    xgb_cv_train_f1.mean(), xgb_cv_train_f1.std()
))
print("Macro F1 (Validation folds): {:.4f} +/- {:.4f}".format(
    xgb_cv_val_f1.mean(), xgb_cv_val_f1.std()
))
print("Accuracy (Validation folds): {:.4f} +/- {:.4f}".format(
    xgb_cv_val_acc.mean(), xgb_cv_val_acc.std()
))

xgb_overfit_gap = xgb_cv_train_f1.mean() - xgb_cv_val_f1.mean()
print("\nTrain-Validation Macro F1 Gap (Overfitting indicator): {:.4f}".format(
    xgb_overfit_gap
))


# ------------------------------------------------------------
# 6. Overfitting interpretation
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("OVERFITTING ASSESSMENT")
print("=" * 70)

def interpret_gap(gap):
    if gap < 0.03:
        return "Low overfitting risk (gap < 0.03)"
    elif gap < 0.08:
        return "Moderate overfitting risk (0.03 <= gap < 0.08)"
    else:
        return "High overfitting risk (gap >= 0.08)"

print("Random Forest gap: {:.4f} -> {}".format(rf_overfit_gap, interpret_gap(rf_overfit_gap)))
print("XGBoost gap      : {:.4f} -> {}".format(xgb_overfit_gap, interpret_gap(xgb_overfit_gap)))


# ------------------------------------------------------------
# 7. Held-out test set vs CV validation consistency check
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("HELD-OUT TEST vs CROSS-VALIDATION CONSISTENCY")
print("=" * 70)

rf_test_vs_cv_gap = abs(rf_macro_f1 - rf_cv_val_f1.mean())
xgb_test_vs_cv_gap = abs(xgb_macro_f1 - xgb_cv_val_f1.mean())

print("Random Forest: Held-out Macro F1 = {:.4f} | CV Val Macro F1 mean = {:.4f} | Diff = {:.4f}".format(
    rf_macro_f1, rf_cv_val_f1.mean(), rf_test_vs_cv_gap
))
print("XGBoost      : Held-out Macro F1 = {:.4f} | CV Val Macro F1 mean = {:.4f} | Diff = {:.4f}".format(
    xgb_macro_f1, xgb_cv_val_f1.mean(), xgb_test_vs_cv_gap
))

if rf_test_vs_cv_gap < 0.05 and xgb_test_vs_cv_gap < 0.05:
    print("\n✓ Held-out test performance is consistent with cross-validation estimates.")
else:
    print("\n⚠ Notable discrepancy between held-out test and CV estimates -- interpret held-out result with caution.")


# ------------------------------------------------------------
# 8. Paired statistical comparison (Wilcoxon signed-rank test)
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("STATISTICAL SIGNIFICANCE: RF vs XGBoost (fold-wise Macro F1)")
print("=" * 70)

wilcoxon_stat, wilcoxon_p = stats.wilcoxon(rf_cv_val_f1, xgb_cv_val_f1)

print("Wilcoxon signed-rank statistic:", round(wilcoxon_stat, 4))
print("p-value                       :", round(wilcoxon_p, 4))

alpha = 0.05
if wilcoxon_p < alpha:
    print(f"✓ Difference is statistically significant at alpha = {alpha}.")
else:
    print(f"⚠ Difference is NOT statistically significant at alpha = {alpha} "
          f"(n_folds is small, interpret cautiously).")


# ------------------------------------------------------------
# 9. Cross-validation results table
# ------------------------------------------------------------

cv_summary = pd.DataFrame([
    {
        "Model": "Random Forest",
        "CV Train Macro F1 (Mean)": rf_cv_train_f1.mean(),
        "CV Train Macro F1 (Std)": rf_cv_train_f1.std(),
        "CV Val Macro F1 (Mean)": rf_cv_val_f1.mean(),
        "CV Val Macro F1 (Std)": rf_cv_val_f1.std(),
        "CV Val Accuracy (Mean)": rf_cv_val_acc.mean(),
        "Train-Val Gap": rf_overfit_gap,
        "Held-out Test Macro F1": rf_macro_f1
    },
    {
        "Model": "XGBoost",
        "CV Train Macro F1 (Mean)": xgb_cv_train_f1.mean(),
        "CV Train Macro F1 (Std)": xgb_cv_train_f1.std(),
        "CV Val Macro F1 (Mean)": xgb_cv_val_f1.mean(),
        "CV Val Macro F1 (Std)": xgb_cv_val_f1.std(),
        "CV Val Accuracy (Mean)": xgb_cv_val_acc.mean(),
        "Train-Val Gap": xgb_overfit_gap,
        "Held-out Test Macro F1": xgb_macro_f1
    }
])

print("\n" + "=" * 70)
print("CROSS-VALIDATION SUMMARY TABLE")
print("=" * 70)

display(cv_summary.round(4))


# ------------------------------------------------------------
# 10. Visualization: fold-wise Macro F1 (Train vs Validation)
# ------------------------------------------------------------

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
folds = np.arange(1, N_SPLITS + 1)

axes[0].plot(folds, rf_cv_train_f1, marker="o", label="Train Macro F1")
axes[0].plot(folds, rf_cv_val_f1, marker="o", label="Validation Macro F1")
axes[0].axhline(rf_macro_f1, color="red", linestyle="--", label="Held-out Test Macro F1")
axes[0].set_title("Random Forest: CV Macro F1 per Fold")
axes[0].set_xlabel("Fold")
axes[0].set_ylabel("Macro F1")
axes[0].set_xticks(folds)
axes[0].legend()
axes[0].grid(alpha=0.3)

axes[1].plot(folds, xgb_cv_train_f1, marker="o", label="Train Macro F1")
axes[1].plot(folds, xgb_cv_val_f1, marker="o", label="Validation Macro F1")
axes[1].axhline(xgb_macro_f1, color="red", linestyle="--", label="Held-out Test Macro F1")
axes[1].set_title("XGBoost: CV Macro F1 per Fold")
axes[1].set_xlabel("Fold")
axes[1].set_ylabel("Macro F1")
axes[1].set_xticks(folds)
axes[1].legend()
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.savefig("cv_fold_performance.png", dpi=150)
plt.show()


# ------------------------------------------------------------
# 11. Boxplot comparison of validation fold scores
# ------------------------------------------------------------

plt.figure(figsize=(8, 6))
plt.boxplot(
    [rf_cv_val_f1, xgb_cv_val_f1],
    tick_labels=["Random Forest", "XGBoost"]
)
plt.ylabel("Validation Macro F1")
plt.title("Cross-Validation Macro F1 Distribution (5-Fold)")
plt.grid(alpha=0.3)
plt.savefig("cv_boxplot_comparison.png", dpi=150)
plt.show()


# ------------------------------------------------------------
# 12. Save results
# ------------------------------------------------------------

cv_summary.round(4).to_csv("cross_validation_results.csv", index=False)

print("\n" + "=" * 70)
print("MODEL VALIDATION / RELIABILITY COMPLETED")
print("=" * 70)
print("✓ 5-Fold Stratified Cross-Validation performed on training set")
print("✓ Macro F1 mean +/- std reported for both models")
print("✓ Overfitting gap (Train - Validation) calculated")
print("✓ Held-out test vs CV consistency checked")
print("✓ Wilcoxon signed-rank test performed (statistical significance)")
print("✓ cross_validation_results.csv saved")
```

```python
# ============================================================
# STEP 14: MODEL VALIDATION / RELIABILITY
# Research-Oriented ML Workflow
#
# IMPROVED FROM ORIGINAL:
#   - StratifiedKFold (5 folds, n=5 samples) -> RepeatedStratifiedKFold
#     (5 folds x 5 repeats = n=25 samples). This gives the Wilcoxon
#     test much more statistical power without touching rf_model /
#     xgb_model at all -- same hyperparameters as your original.
# ============================================================

from sklearn.model_selection import RepeatedStratifiedKFold
from sklearn.metrics import f1_score, accuracy_score
from sklearn.base import clone
from sklearn.utils.class_weight import compute_sample_weight
from scipy import stats
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("=" * 70)
print("STEP 14: MODEL VALIDATION / RELIABILITY")
print("=" * 70)


# ------------------------------------------------------------
# 1. Define Repeated Stratified K-Fold cross-validation strategy
# ------------------------------------------------------------

N_SPLITS = 5
N_REPEATS = 5   # 5 folds x 5 repeats = 25 paired samples (was 5 before)

cv = RepeatedStratifiedKFold(
    n_splits=N_SPLITS,
    n_repeats=N_REPEATS,
    random_state=42
)

TOTAL_RUNS = N_SPLITS * N_REPEATS

print("\nCross-validation strategy : Repeated Stratified K-Fold")
print("Splits x Repeats          : {} x {} = {} runs".format(N_SPLITS, N_REPEATS, TOTAL_RUNS))
print("Performed on               : Training set only (X_train, y_train)")
print("Test set                   : Untouched (held out)")


# ------------------------------------------------------------
# 2. Random Forest -- Repeated Stratified K-Fold cross-validation
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("RANDOM FOREST -- CROSS-VALIDATION")
print("=" * 70)

rf_cv_train_f1 = []
rf_cv_val_f1 = []
rf_cv_val_acc = []

for run, (train_idx, val_idx) in enumerate(cv.split(X_train, y_train), start=1):

    X_tr_fold = X_train.iloc[train_idx]
    y_tr_fold = y_train.iloc[train_idx]

    X_val_fold = X_train.iloc[val_idx]
    y_val_fold = y_train.iloc[val_idx]

    rf_fold_model = clone(rf_model)
    rf_fold_model.fit(X_tr_fold, y_tr_fold)

    # Performance on the fold's own training data
    y_tr_pred = rf_fold_model.predict(X_tr_fold)
    train_f1 = f1_score(y_tr_fold, y_tr_pred, average="macro", zero_division=0)

    # Performance on the held-out validation fold
    y_val_pred = rf_fold_model.predict(X_val_fold)
    val_f1 = f1_score(y_val_fold, y_val_pred, average="macro", zero_division=0)
    val_acc = accuracy_score(y_val_fold, y_val_pred)

    rf_cv_train_f1.append(train_f1)
    rf_cv_val_f1.append(val_f1)
    rf_cv_val_acc.append(val_acc)

    print(
        f"Run {run}/{TOTAL_RUNS}: "
        f"Train Macro F1 = {train_f1:.4f} | "
        f"Val Macro F1 = {val_f1:.4f} | "
        f"Val Accuracy = {val_acc:.4f}"
    )

rf_cv_train_f1 = np.array(rf_cv_train_f1)
rf_cv_val_f1 = np.array(rf_cv_val_f1)
rf_cv_val_acc = np.array(rf_cv_val_acc)


# ------------------------------------------------------------
# 3. Random Forest -- cross-validation summary
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("RANDOM FOREST -- CROSS-VALIDATION SUMMARY")
print("=" * 70)

print("Macro F1 (Train, n={})     : {:.4f} +/- {:.4f}".format(
    len(rf_cv_train_f1), rf_cv_train_f1.mean(), rf_cv_train_f1.std()
))
print("Macro F1 (Validation, n={}): {:.4f} +/- {:.4f}".format(
    len(rf_cv_val_f1), rf_cv_val_f1.mean(), rf_cv_val_f1.std()
))
print("Accuracy (Validation)      : {:.4f} +/- {:.4f}".format(
    rf_cv_val_acc.mean(), rf_cv_val_acc.std()
))

rf_overfit_gap = rf_cv_train_f1.mean() - rf_cv_val_f1.mean()
print("\nTrain-Validation Macro F1 Gap (Overfitting indicator): {:.4f}".format(
    rf_overfit_gap
))


# ------------------------------------------------------------
# 4. XGBoost -- Repeated Stratified K-Fold cross-validation
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("XGBOOST -- CROSS-VALIDATION")
print("=" * 70)

xgb_cv_train_f1 = []
xgb_cv_val_f1 = []
xgb_cv_val_acc = []

# Encode the full training target once (consistent with label_encoder from Step 10)
y_train_encoded_full = label_encoder.transform(y_train)

for run, (train_idx, val_idx) in enumerate(cv.split(X_train, y_train), start=1):

    X_tr_fold = X_train.iloc[train_idx]
    y_tr_fold_enc = y_train_encoded_full[train_idx]

    X_val_fold = X_train.iloc[val_idx]
    y_val_fold_enc = y_train_encoded_full[val_idx]

    # Balanced sample weights computed ONLY from this fold's training data
    fold_sample_weights = compute_sample_weight(
        class_weight="balanced",
        y=y_tr_fold_enc
    )

    xgb_fold_model = clone(xgb_model)
    xgb_fold_model.fit(
        X_tr_fold,
        y_tr_fold_enc,
        sample_weight=fold_sample_weights
    )

    # Performance on the fold's own training data
    y_tr_pred_enc = xgb_fold_model.predict(X_tr_fold)
    train_f1 = f1_score(y_tr_fold_enc, y_tr_pred_enc, average="macro", zero_division=0)

    # Performance on the held-out validation fold
    y_val_pred_enc = xgb_fold_model.predict(X_val_fold)
    val_f1 = f1_score(y_val_fold_enc, y_val_pred_enc, average="macro", zero_division=0)
    val_acc = accuracy_score(y_val_fold_enc, y_val_pred_enc)

    xgb_cv_train_f1.append(train_f1)
    xgb_cv_val_f1.append(val_f1)
    xgb_cv_val_acc.append(val_acc)

    print(
        f"Run {run}/{TOTAL_RUNS}: "
        f"Train Macro F1 = {train_f1:.4f} | "
        f"Val Macro F1 = {val_f1:.4f} | "
        f"Val Accuracy = {val_acc:.4f}"
    )

xgb_cv_train_f1 = np.array(xgb_cv_train_f1)
xgb_cv_val_f1 = np.array(xgb_cv_val_f1)
xgb_cv_val_acc = np.array(xgb_cv_val_acc)


# ------------------------------------------------------------
# 5. XGBoost -- cross-validation summary
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("XGBOOST -- CROSS-VALIDATION SUMMARY")
print("=" * 70)

print("Macro F1 (Train, n={})     : {:.4f} +/- {:.4f}".format(
    len(xgb_cv_train_f1), xgb_cv_train_f1.mean(), xgb_cv_train_f1.std()
))
print("Macro F1 (Validation, n={}): {:.4f} +/- {:.4f}".format(
    len(xgb_cv_val_f1), xgb_cv_val_f1.mean(), xgb_cv_val_f1.std()
))
print("Accuracy (Validation)      : {:.4f} +/- {:.4f}".format(
    xgb_cv_val_acc.mean(), xgb_cv_val_acc.std()
))

xgb_overfit_gap = xgb_cv_train_f1.mean() - xgb_cv_val_f1.mean()
print("\nTrain-Validation Macro F1 Gap (Overfitting indicator): {:.4f}".format(
    xgb_overfit_gap
))


# ------------------------------------------------------------
# 6. Overfitting interpretation
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("OVERFITTING ASSESSMENT")
print("=" * 70)

def interpret_gap(gap):
    if gap < 0.03:
        return "Low overfitting risk (gap < 0.03)"
    elif gap < 0.08:
        return "Moderate overfitting risk (0.03 <= gap < 0.08)"
    else:
        return "High overfitting risk (gap >= 0.08)"

print("Random Forest gap: {:.4f} -> {}".format(rf_overfit_gap, interpret_gap(rf_overfit_gap)))
print("XGBoost gap      : {:.4f} -> {}".format(xgb_overfit_gap, interpret_gap(xgb_overfit_gap)))


# ------------------------------------------------------------
# 7. Held-out test set vs CV validation consistency check
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("HELD-OUT TEST vs CROSS-VALIDATION CONSISTENCY")
print("=" * 70)

rf_test_vs_cv_gap = abs(rf_macro_f1 - rf_cv_val_f1.mean())
xgb_test_vs_cv_gap = abs(xgb_macro_f1 - xgb_cv_val_f1.mean())

print("Random Forest: Held-out Macro F1 = {:.4f} | CV Val Macro F1 mean = {:.4f} | Diff = {:.4f}".format(
    rf_macro_f1, rf_cv_val_f1.mean(), rf_test_vs_cv_gap
))
print("XGBoost      : Held-out Macro F1 = {:.4f} | CV Val Macro F1 mean = {:.4f} | Diff = {:.4f}".format(
    xgb_macro_f1, xgb_cv_val_f1.mean(), xgb_test_vs_cv_gap
))

if rf_test_vs_cv_gap < 0.05 and xgb_test_vs_cv_gap < 0.05:
    print("\n✓ Held-out test performance is consistent with cross-validation estimates.")
else:
    print("\n⚠ Notable discrepancy between held-out test and CV estimates -- interpret held-out result with caution.")


# ------------------------------------------------------------
# 8. Paired statistical comparison (Wilcoxon signed-rank test)
#    Now with n=25 paired observations instead of n=5
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("STATISTICAL SIGNIFICANCE: RF vs XGBoost (n={} paired runs)".format(TOTAL_RUNS))
print("=" * 70)

wilcoxon_stat, wilcoxon_p = stats.wilcoxon(rf_cv_val_f1, xgb_cv_val_f1)

print("Wilcoxon signed-rank statistic:", round(wilcoxon_stat, 4))
print("p-value                       :", round(wilcoxon_p, 4))

alpha = 0.05
if wilcoxon_p < alpha:
    print(f"✓ Difference is statistically significant at alpha = {alpha}.")
else:
    print(f"⚠ Difference is NOT statistically significant at alpha = {alpha} "
          f"(n={TOTAL_RUNS} paired runs, interpret with normal caution).")


# ------------------------------------------------------------
# 9. Cross-validation results table
# ------------------------------------------------------------

cv_summary = pd.DataFrame([
    {
        "Model": "Random Forest",
        "CV Train Macro F1 (Mean)": rf_cv_train_f1.mean(),
        "CV Train Macro F1 (Std)": rf_cv_train_f1.std(),
        "CV Val Macro F1 (Mean)": rf_cv_val_f1.mean(),
        "CV Val Macro F1 (Std)": rf_cv_val_f1.std(),
        "CV Val Accuracy (Mean)": rf_cv_val_acc.mean(),
        "Train-Val Gap": rf_overfit_gap,
        "Held-out Test Macro F1": rf_macro_f1,
        "N Runs": len(rf_cv_val_f1)
    },
    {
        "Model": "XGBoost",
        "CV Train Macro F1 (Mean)": xgb_cv_train_f1.mean(),
        "CV Train Macro F1 (Std)": xgb_cv_train_f1.std(),
        "CV Val Macro F1 (Mean)": xgb_cv_val_f1.mean(),
        "CV Val Macro F1 (Std)": xgb_cv_val_f1.std(),
        "CV Val Accuracy (Mean)": xgb_cv_val_acc.mean(),
        "Train-Val Gap": xgb_overfit_gap,
        "Held-out Test Macro F1": xgb_macro_f1,
        "N Runs": len(xgb_cv_val_f1)
    }
])

print("\n" + "=" * 70)
print("CROSS-VALIDATION SUMMARY TABLE")
print("=" * 70)

display(cv_summary.round(4))


# ------------------------------------------------------------
# 10. Visualization: Macro F1 per run (Train vs Validation)
# ------------------------------------------------------------

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
runs = np.arange(1, TOTAL_RUNS + 1)

axes[0].plot(runs, rf_cv_train_f1, marker=".", alpha=0.7, label="Train Macro F1")
axes[0].plot(runs, rf_cv_val_f1, marker=".", alpha=0.7, label="Validation Macro F1")
axes[0].axhline(rf_macro_f1, color="red", linestyle="--", label="Held-out Test Macro F1")
axes[0].set_title("Random Forest: Macro F1 per Run ({} runs)".format(TOTAL_RUNS))
axes[0].set_xlabel("Run")
axes[0].set_ylabel("Macro F1")
axes[0].legend()
axes[0].grid(alpha=0.3)

axes[1].plot(runs, xgb_cv_train_f1, marker=".", alpha=0.7, label="Train Macro F1")
axes[1].plot(runs, xgb_cv_val_f1, marker=".", alpha=0.7, label="Validation Macro F1")
axes[1].axhline(xgb_macro_f1, color="red", linestyle="--", label="Held-out Test Macro F1")
axes[1].set_title("XGBoost: Macro F1 per Run ({} runs)".format(TOTAL_RUNS))
axes[1].set_xlabel("Run")
axes[1].set_ylabel("Macro F1")
axes[1].legend()
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.savefig("cv_fold_performance.png", dpi=150)
plt.show()


# ------------------------------------------------------------
# 11. Boxplot comparison of validation fold scores
# ------------------------------------------------------------

plt.figure(figsize=(8, 6))
plt.boxplot(
    [rf_cv_val_f1, xgb_cv_val_f1],
    tick_labels=["Random Forest", "XGBoost"]
)
plt.ylabel("Validation Macro F1")
plt.title("Cross-Validation Macro F1 Distribution (n={} each)".format(TOTAL_RUNS))
plt.grid(alpha=0.3)
plt.savefig("cv_boxplot_comparison.png", dpi=150)
plt.show()


# ------------------------------------------------------------
# 12. Save results
# ------------------------------------------------------------

cv_summary.round(4).to_csv("cross_validation_results.csv", index=False)

print("\n" + "=" * 70)
print("MODEL VALIDATION / RELIABILITY COMPLETED")
print("=" * 70)
print(f"✓ Repeated Stratified K-Fold CV performed ({N_SPLITS} splits x {N_REPEATS} repeats = {TOTAL_RUNS} runs)")
print("✓ Macro F1 mean +/- std reported for both models")
print("✓ Overfitting gap (Train - Validation) calculated")
print("✓ Held-out test vs CV consistency checked")
print(f"✓ Wilcoxon signed-rank test performed with n={TOTAL_RUNS} (was n=5 before)")
print("✓ cross_validation_results.csv saved")
```

```python
# ============================================================
# STEP 15: FEATURE IMPORTANCE COMPARISON
# Research-Oriented Explainable AI (XAI)
#
# Compares three different feature importance methods:
#   1. Random Forest built-in importance (Mean Decrease in Impurity / Gini)
#   2. XGBoost built-in importance (gain-based)
#   3. SHAP importance (Mean Absolute SHAP, from Step 13, best model)
#
# Note: RF and XGBoost built-in importance measure how much each
# feature reduces impurity/loss DURING TRAINING. SHAP importance
# measures each feature's actual contribution to individual
# PREDICTIONS on the test set. They can disagree -- that is a
# meaningful research finding, not an error.
# ============================================================

import numpy as np
import pandas as pd
from scipy.stats import spearmanr
import matplotlib.pyplot as plt

print("=" * 70)
print("STEP 15: FEATURE IMPORTANCE COMPARISON")
print("=" * 70)


# ------------------------------------------------------------
# 1. Random Forest feature importance (Mean Decrease in Impurity)
# ------------------------------------------------------------

rf_importance = pd.DataFrame({
    "Feature": X_train.columns,
    "RF Importance": rf_model.feature_importances_
})

print("\n" + "=" * 70)
print("RANDOM FOREST FEATURE IMPORTANCE")
print("=" * 70)
display(rf_importance.sort_values("RF Importance", ascending=False).reset_index(drop=True))


# ------------------------------------------------------------
# 2. XGBoost feature importance (gain-based)
# ------------------------------------------------------------

xgb_importance = pd.DataFrame({
    "Feature": X_train.columns,
    "XGBoost Importance": xgb_model.feature_importances_
})

print("\n" + "=" * 70)
print("XGBOOST FEATURE IMPORTANCE")
print("=" * 70)
display(xgb_importance.sort_values("XGBoost Importance", ascending=False).reset_index(drop=True))


# ------------------------------------------------------------
# 3. SHAP importance
#    Rebuilt directly from shap_values_array (computed in Step 13)
#    rather than reusing global_shap_importance, since that
#    variable's type/name can vary depending on how Step 13 was
#    written. This only needs shap_values_array + shap_data to exist.
#    If Step 13 has not been run yet, this cell will raise a
#    NameError -- run Step 13 first.
# ------------------------------------------------------------

if shap_values_array.ndim == 3:
    # multiclass: average absolute SHAP over samples AND classes
    global_shap_importance_values = np.abs(shap_values_array).mean(axis=(0, 2))
else:
    # binary / single-output: average absolute SHAP over samples
    global_shap_importance_values = np.abs(shap_values_array).mean(axis=0)

shap_importance = pd.DataFrame({
    "Feature": shap_data.columns,
    "SHAP Importance": global_shap_importance_values
}).sort_values("SHAP Importance", ascending=False).reset_index(drop=True)

print(f"\n" + "=" * 70)
print(f"SHAP FEATURE IMPORTANCE (from {shap_model_name}, rebuilt from Step 13's shap_values_array)")
print("=" * 70)
display(shap_importance)


# ------------------------------------------------------------
# 4. Merge all three into a single comparison table
#    Normalize each column to sum to 1 so all three methods
#    are on the same scale and directly comparable
# ------------------------------------------------------------

comparison = (
    rf_importance
    .merge(xgb_importance, on="Feature")
    .merge(shap_importance, on="Feature")
)

for col in ["RF Importance", "XGBoost Importance", "SHAP Importance"]:
    comparison[col + " (Normalized)"] = comparison[col] / comparison[col].sum()

comparison_sorted = comparison.sort_values(
    "SHAP Importance (Normalized)", ascending=False
).reset_index(drop=True)

print("\n" + "=" * 70)
print("FEATURE IMPORTANCE COMPARISON TABLE (Normalized, sums to 1 per method)")
print("=" * 70)
display(comparison_sorted[[
    "Feature",
    "RF Importance (Normalized)",
    "XGBoost Importance (Normalized)",
    "SHAP Importance (Normalized)"
]].round(4))


# ------------------------------------------------------------
# 5. Rank agreement between methods (Spearman correlation)
#    High rho = methods rank features similarly.
#    Low/near-zero rho = methods disagree on what matters.
# ------------------------------------------------------------

rf_vs_xgb_corr, rf_vs_xgb_p = spearmanr(comparison["RF Importance"], comparison["XGBoost Importance"])
rf_vs_shap_corr, rf_vs_shap_p = spearmanr(comparison["RF Importance"], comparison["SHAP Importance"])
xgb_vs_shap_corr, xgb_vs_shap_p = spearmanr(comparison["XGBoost Importance"], comparison["SHAP Importance"])

print("\n" + "=" * 70)
print("RANK AGREEMENT BETWEEN METHODS (Spearman correlation)")
print("=" * 70)
print("RF vs XGBoost   : rho = {:.4f} (p = {:.4f})".format(rf_vs_xgb_corr, rf_vs_xgb_p))
print("RF vs SHAP      : rho = {:.4f} (p = {:.4f})".format(rf_vs_shap_corr, rf_vs_shap_p))
print("XGBoost vs SHAP : rho = {:.4f} (p = {:.4f})".format(xgb_vs_shap_corr, xgb_vs_shap_p))

print("\nInterpretation guide:")
print("  rho > 0.7  -> strong agreement")
print("  rho 0.4-0.7 -> moderate agreement")
print("  rho < 0.4  -> weak agreement (methods disagree on feature ranking)")


# ------------------------------------------------------------
# 6. Top-3 features per method + consensus features
# ------------------------------------------------------------

top3_rf = set(rf_importance.sort_values("RF Importance", ascending=False)["Feature"].head(3))
top3_xgb = set(xgb_importance.sort_values("XGBoost Importance", ascending=False)["Feature"].head(3))
top3_shap = set(shap_importance.sort_values("SHAP Importance", ascending=False)["Feature"].head(3))

consensus_top3 = top3_rf & top3_xgb & top3_shap
union_top3 = top3_rf | top3_xgb | top3_shap

print("\n" + "=" * 70)
print("TOP-3 FEATURES PER METHOD")
print("=" * 70)
print("Random Forest top-3:", top3_rf)
print("XGBoost top-3      :", top3_xgb)
print("SHAP top-3         :", top3_shap)

print("\nFeatures appearing in ALL THREE top-3 lists (strong consensus):")
print(" ", consensus_top3 if consensus_top3 else "None -- no feature is top-3 across all methods")

print("\nFeatures appearing in AT LEAST ONE top-3 list:")
print(" ", union_top3)


# ------------------------------------------------------------
# 7. Visualization: grouped horizontal bar chart, all 3 methods
# ------------------------------------------------------------

plot_df = comparison_sorted.set_index("Feature")[[
    "RF Importance (Normalized)",
    "XGBoost Importance (Normalized)",
    "SHAP Importance (Normalized)"
]]
plot_df.columns = ["Random Forest", "XGBoost", "SHAP"]

ax = plot_df.plot(kind="barh", figsize=(10, 7), width=0.75)
ax.set_xlabel("Normalized Importance")
ax.set_title("Feature Importance Comparison: Random Forest vs XGBoost vs SHAP")
ax.invert_yaxis()
ax.legend(title="Method")
plt.tight_layout()
plt.savefig("feature_importance_comparison.png", dpi=150)
plt.show()


# ------------------------------------------------------------
# 8. Save comparison table
# ------------------------------------------------------------

comparison_sorted.round(4).to_csv("feature_importance_comparison.csv", index=False)

print("\n" + "=" * 70)
print("FEATURE IMPORTANCE COMPARISON COMPLETED")
print("=" * 70)
print("✓ RF, XGBoost, and SHAP importance computed and compared")
print("✓ Spearman rank correlation calculated between all method pairs")
print("✓ Top-3 consensus features identified")
print("✓ feature_importance_comparison.png saved")
print("✓ feature_importance_comparison.csv saved")
```

```python
# ============================================================
# STEP 17: INDIVIDUAL DISEASE PREDICTION
# Research-Oriented ML Workflow
#
# Takes new (unseen) environmental sensor readings, applies the
# SAME preprocessing pipeline used during training, and returns:
#   - Predicted disease (Random Forest & XGBoost)
#   - Full class-wise prediction probability distribution
#   - Model agreement check
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("=" * 70)
print("STEP 17: INDIVIDUAL DISEASE PREDICTION")
print("=" * 70)


# ------------------------------------------------------------
# 1. Reusable prediction function
# ------------------------------------------------------------
def predict_disease(new_data: dict, verbose=True):
    """
    Parameters
    ----------
    new_data : dict
        Raw environmental/IoT sensor reading, e.g.
        {
            "crop_type": "Tomato",
            "season": "winter",
            "temperature": 21.5,
            "humidity": 68.0,
            "rainfall": 2.5,
            "soil_moisture": 48.0,
            "soil_ph": 6.1,
            "light_intensity": 42000
        }

    Returns
    -------
    dict with predictions + probability tables from both models
    """

    # --- 1a. Convert single sample to DataFrame ---
    new_df = pd.DataFrame([new_data])

    # --- 1b. Encode categorical features using the SAME fitted encoder ---
    new_encoded_array = encoder.transform(new_df[categorical_features])
    new_encoded_df = pd.DataFrame(
        new_encoded_array,
        columns=encoder.get_feature_names_out(categorical_features),
        index=new_df.index
    )

    # --- 1c. Combine numerical + encoded categorical (same order as training) ---
    new_X = pd.concat(
        [new_df[numerical_features], new_encoded_df],
        axis=1
    )

    # --- 1d. Align columns exactly with training feature matrix X ---
    new_X = new_X.reindex(columns=X.columns, fill_value=0)

    # ------------------------------------------------------------
    # 2. Random Forest prediction + probabilities
    # ------------------------------------------------------------
    rf_pred = rf_model.predict(new_X)[0]
    rf_proba = rf_model.predict_proba(new_X)[0]

    rf_proba_table = pd.DataFrame({
        "Disease": rf_model.classes_,
        "Probability": rf_proba
    }).sort_values("Probability", ascending=False).reset_index(drop=True)

    # ------------------------------------------------------------
    # 3. XGBoost prediction + probabilities
    # ------------------------------------------------------------
    xgb_pred_encoded = xgb_model.predict(new_X)[0]
    xgb_pred = label_encoder.inverse_transform([xgb_pred_encoded])[0]
    xgb_proba = xgb_model.predict_proba(new_X)[0]

    xgb_proba_table = pd.DataFrame({
        "Disease": label_encoder.classes_,
        "Probability": xgb_proba
    }).sort_values("Probability", ascending=False).reset_index(drop=True)

    # ------------------------------------------------------------
    # 4. Display results
    # ------------------------------------------------------------
    if verbose:
        print("\n" + "=" * 70)
        print("NEW ENVIRONMENTAL DATA INPUT")
        print("=" * 70)
        display(new_df)

        print("\n" + "=" * 70)
        print("RANDOM FOREST PREDICTION")
        print("=" * 70)
        print(f"Predicted Disease : {rf_pred}")
        print(f"Confidence        : {rf_proba.max():.4f} ({rf_proba.max()*100:.2f}%)")
        print("\nFull probability distribution:")
        display(rf_proba_table.round(4))

        print("\n" + "=" * 70)
        print("XGBOOST PREDICTION")
        print("=" * 70)
        print(f"Predicted Disease : {xgb_pred}")
        print(f"Confidence        : {xgb_proba.max():.4f} ({xgb_proba.max()*100:.2f}%)")
        print("\nFull probability distribution:")
        display(xgb_proba_table.round(4))

        print("\n" + "=" * 70)
        print("MODEL AGREEMENT CHECK")
        print("=" * 70)
        if rf_pred == xgb_pred:
            print(f"✓ Both models agree: {rf_pred}")
        else:
            print(f"⚠ Models disagree -> RF: {rf_pred} | XGBoost: {xgb_pred}")

        # --- Visualization: side-by-side probability bar chart ---
        fig, axes = plt.subplots(1, 2, figsize=(14, 5))

        axes[0].barh(rf_proba_table["Disease"], rf_proba_table["Probability"], color="#4C72B0")
        axes[0].set_xlim(0, 1)
        axes[0].set_xlabel("Probability")
        axes[0].set_title("Random Forest - Prediction Probability")
        axes[0].invert_yaxis()

        axes[1].barh(xgb_proba_table["Disease"], xgb_proba_table["Probability"], color="#DD8452")
        axes[1].set_xlim(0, 1)
        axes[1].set_xlabel("Probability")
        axes[1].set_title("XGBoost - Prediction Probability")
        axes[1].invert_yaxis()

        plt.tight_layout()
        plt.show()

    return {
        "input": new_data,
        "rf_prediction": rf_pred,
        "rf_confidence": rf_proba.max(),
        "rf_probabilities": rf_proba_table,
        "xgb_prediction": xgb_pred,
        "xgb_confidence": xgb_proba.max(),
        "xgb_probabilities": xgb_proba_table
    }


# ------------------------------------------------------------
# 5. Example 1: New unseen sample
# ------------------------------------------------------------
sample_1 = {
    "crop_type": "Tomato",
    "season": "winter",
    "temperature": 21.5,
    "humidity": 68.0,
    "rainfall": 2.5,
    "soil_moisture": 48.0,
    "soil_ph": 6.1,
    "light_intensity": 42000
}

print("\n" + "#" * 70)
print("# EXAMPLE 1")
print("#" * 70)
result_1 = predict_disease(sample_1)


# ------------------------------------------------------------
# 6. Example 2: Another new sample (different crop/season)
# ------------------------------------------------------------
sample_2 = {
    "crop_type": "Potato",
    "season": "summer",
    "temperature": 29.0,
    "humidity": 40.0,
    "rainfall": 0.0,
    "soil_moisture": 30.0,
    "soil_ph": 5.5,
    "light_intensity": 55000
}

print("\n" + "#" * 70)
print("# EXAMPLE 2")
print("#" * 70)
result_2 = predict_disease(sample_2)


# ------------------------------------------------------------
# 7. Summary table of individual predictions
# ------------------------------------------------------------
summary_table = pd.DataFrame([
    {
        "Sample": "Sample 1 (Tomato, Winter)",
        "RF Prediction": result_1["rf_prediction"],
        "RF Confidence": round(result_1["rf_confidence"], 4),
        "XGB Prediction": result_1["xgb_prediction"],
        "XGB Confidence": round(result_1["xgb_confidence"], 4),
    },
    {
        "Sample": "Sample 2 (Potato, Summer)",
        "RF Prediction": result_2["rf_prediction"],
        "RF Confidence": round(result_2["rf_confidence"], 4),
        "XGB Prediction": result_2["xgb_prediction"],
        "XGB Confidence": round(result_2["xgb_confidence"], 4),
    }
])

print("\n" + "=" * 70)
print("INDIVIDUAL PREDICTION SUMMARY")
print("=" * 70)
display(summary_table)

summary_table.to_csv("individual_predictions.csv", index=False)
print("\n✓ individual_predictions.csv saved")
```
